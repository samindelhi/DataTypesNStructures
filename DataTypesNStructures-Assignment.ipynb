8. What is a hash table, and how does it relate to dictionaries in Python
    - A hash table is a data structure that maps keys to values using a hashing function, ensuring fast lookups, insertions, and deletions.
    - Python’s dictionaries (dict) are implemented using hash tables, making key-based access highly efficient.
    - Hash Tables Work in Dictionaries provide:
        - Hashing: Each key is processed using a hash function, which converts it into -a unique numerical index.
        - Index-Based Storage: The hash value determines where the key-value pair is stored.
        - Fast Lookups (0 Complexity): Since dictionaries use hashing, retrieval is nearly instantaneous.

17. Describe the time complexity of accessing elements in a dictionary
- Accessing elements in a dictionary is highly efficient due to the hashing mechanism used for key-value storage. Here’s how time complexity plays a role:
Time Complexity of Dictionary Access for-
- **Average Case**: 0 (Constant Time)- Dictionary lookups are performed using hashing, meaning the key is mapped to a memory location, allowing direct access.
    - Example:my_dict = {"apple": 100, "banana": 50, "cherry": 200}
    - price = my_dict["banana"]  # O
- **Worst case**: 0 (Linear time) - If hash collisions occur (multiple keys mapping to the same hash index), Python must resolve them, possibly by searching through a linked list or rehashing. This is rare but can happen if a poor hashing function distributes keys inefficiently

18. In what situations are lists preferred over dictionaries
- Lists are preferred over dictionaries in situations where ordered, indexed, or sequential data access is important. They are also preferred where sorting, storing duplicates, managing simple collections without any key-value mapping.

19. Why are dictionaries considered unordered, and how does that affect data retrieval
- Dictionaries in Python are considered unordered because they store key-value pairs using a hashing mechanism, not a fixed sequence. Due to this the key lookups are incredibly fast compared to the list. Dictionaries were Unordered before Python 3.7, and are now insertion-ordered since then.

```python
# 29. Write a code to combine two dictionaries into one and print the result.
dict1 = {"id":1234,"name":"John","age":34,"dept":"Finance"}
dict2 = {"dept":"Accounts","role":"Manager"}
# Create a copy of dict1 to preserve the original
mergeddict = dict1.copy() 
mergeddict.update(dict2) # The update() method modifies the dictionary in place, so creating a copy ensures that dict1 remains unchanged.
print(dict1) #Original dictionary
print(mergeddict) #merged dictionary


# -------one more option that is more elegant-------------
mergeddict1 = {**dict1, **dict2} # This method ({**dict1, **dict2}) uses dictionary unpacking to combine them efficiently. If there are overlapping keys, the values in dict2 will overwrite those in dict1.
print(mergeddict1)
```
